**Q2. What does a positive, negative, and zero covariance indicate about the relationship between two variables?**

**Ans-**

Covariance is a statistical measure used to determine the direction of the relationship between two variables. The sign of covariance (positive, negative, or zero) provides information about how the variables move with respect to each other.

* **Positive Covariance**

A positive covariance indicates that the two variables tend to move in the same direction. This means that when the value of one variable increases, the value of the other variable also increases, and when one decreases, the other also decreases.

**Example**:
Income and expenditure generally have positive covariance.

* **Negative Covariance**

A negative covariance indicates that the two variables move in opposite directions. When the value of one variable increases, the value of the other variable tends to decrease, and vice versa.

**Example**:
Price of a product and its demand usually show negative covariance.

* **Zero Covariance**


A zero covariance indicates that there is no linear relationship between the two variables. Changes in one variable do not show any consistent pattern with changes in the other variable.

**Example:**

Shoe size and intelligence typically have zero covariance.



* Positive covariance shows direct relationship

* Negative covariance shows inverse relationship

* Zero covariance indicates no linear association

**Q3. Discuss the limitations of covariance as a measure of relationship between two variables. Why is correlation
preferred in many cases**

**Ans-**

Covariance is a useful statistical measure for identifying the direction of the relationship between two variables. However, it has several limitations that reduce its effectiveness as a measure of relationship. Because of these limitations, correlation is often preferred in many practical applications.


1.   Depends on the Scale of Measurement
The value of covariance is affected by the units in which variables are measured. A change in units (for example, from meters to centimeters) can significantly change the magnitude of covariance, making interpretation difficult.

2.  No Fixed Range
Covariance does not have a defined range of values. It can take any positive or negative value, which makes it hard to judge the strength of the relationship between variables.

3.  Difficult Interpretation
Covariance only indicates the direction of the relationship (positive or negative) but does not provide information about how strong the relationship is.

4.  Not Suitable for Comparison
Since covariance is not standardized, it is difficult to compare relationships across different datasets or between different pairs of variables.


**Why Correlation is Preferred**-

* Correlation overcomes the above limitations of covariance:

* It is unit-free and independent of the scale of measurement.

* Its values lie between –1 and +1, making interpretation easy.

* It measures both the direction and strength of the relationship.

* It allows for easy comparison across different datasets.


Although covariance helps in understanding the direction of a relationship between two variables, its dependence on measurement units and lack of standardization limit its usefulness. Correlation is preferred because it provides a clear, standardized, and interpretable measure of the relationship between variables.


**Q4. Explain the difference between Pearson’s correlation coefficient and Spearman’s rank correlation
coefficient. When would you prefer to use Spearman’s correlation?**

**Ans-**

**1.Pearson’s Correlation Coefficient**

Pearson’s correlation coefficient measures the linear relationship between two continuous variables. It is calculated using the actual numerical values of the data.

**Key Features:**

* Assumes a linear relationship between variables

* Requires data to be normally distributed

* Sensitive to outliers

* Measures both strength and direction of linear association

* Values range from –1 to +1

**2.Spearman’s Rank Correlation Coefficient**

Spearman’s rank correlation coefficient measures the monotonic relationship between two variables using their ranks instead of actual values.

**Key Features:**

* Based on ranks, not raw data

* Does not require normal distribution

* Less affected by outliers

* Measures monotonic (not necessarily linear) relationships

* Values range from –1 to +1

**Q5. If the correlation coefficient between two variables X and Y is 0.85, interpret this value in context. Can you
infer causation from this value? Why or why not?**

**Ans-**

A correlation coefficient of 0.85 between two variables X and Y indicates a strong positive linear relationship between them. This means that as the value of variable X increases, the value of variable Y also tends to increase, and the two variables move closely together in the same direction. The value 0.85, being close to +1, suggests a high degree of association between the variables.

However, causation cannot be inferred from this correlation value. Correlation only measures the strength and direction of association, not a cause-and-effect relationship. There may be other factors, known as lurking or confounding variables, that influence both X and Y. Additionally, it is possible that Y causes X, or that the relationship is coincidental.



While a correlation coefficient of 0.85 indicates a strong positive relationship between X and Y, it does not imply that changes in X cause changes in Y. Establishing causation requires controlled experiments or additional evidence beyond correlation analysis.
