**Q1. What is a null hypothesis (H₀) and why is it important in hypothesis testing?**

**Ans-**


The null hypothesis (H₀) is a statement that there is no effect, no difference, or no relationship between variables in a population.
It represents the status quo or a claim that any observed difference is due to random chance.


* **Importants**



1.   Starting point of hypothesis testing

      All statistical tests begin by assuming the null hypothesis is true.

2.  Provides a basis for decision-making

    Data is used to decide whether there is enough evidence to reject H₀ or fail to reject H₀.

3.  Controls subjectivity

    It prevents us from making conclusions based only on intuition or bias.

4.  Helps measure statistical significance

    Using H₀, we calculate p-values or test statistics to judge whether results are likely due to chance.



**Question 2: What does the significance level (α) represent in hypothesis testing?**

**Ans-**

 The significance level (α) is the maximum probability of making a Type
 error, which means rejecting a true null hypothesis.

* It is chosen before performing the test

* Common values are 0.05, 0.01, and 0.10

* If the p-value ≤ α, the null hypothesis is rejected

**Question 6: When should you use a Z-test instead of a t-test?**

**Ans-**

* A Z-test is used in hypothesis testing when certain conditions related to the sample size and population information are satisfied. It is preferred over a t-test in the following situations:

* The population standard deviation (σ) is known.
When the true population standard deviation is available, the Z-test provides accurate results.

* The sample size is large (n ≥ 30).
For large samples, the sampling distribution of the sample mean approaches a normal distribution due to the Central Limit Theorem, making the Z-test appropriate.

* The data is assumed to be normally distributed or the sample size is sufficiently large to assume normality.

* In contrast, a t-test is used when the population standard deviation is unknown and the sample size is small (n < 30).



* A Z-test should be used instead of a t-test when the population standard deviation is known and the sample size is large. Under these conditions, the Z-test gives reliable and accurate results.
