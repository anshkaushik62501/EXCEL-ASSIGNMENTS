**5.What is Kurtosis and their type?**

**Ans-**

**Kurtosis is a statistical measure that describes the shape of a distribution, specifically the peakedness or flatness of the data relative to a normal distribution.**

**It tells us how sharply data values are clustered around the mean and how heavy the tails of the distribution are.**


**Positive Kurtosis (+ve)**

* Also called Leptokurtic

* The distribution has a high, sharp peak

* The tails are heavy

* More chances of extreme values (outliers)

* Excess kurtosis > 0


**Negative Kurtosis (–ve)**

* Also called Platykurtic

* The distribution has a flat, broad peak

* The tails are light

* Fewer extreme values

* Excess kurtosis < 0


**Zero Kurtosis**

* Also called Mesokurtic

* Shape is similar to a normal distribution

* Excess kurtosis = 0

**6.Explain why the uniform distribution is a good model for the outcome of rolling a fair die.**

**Ans-**




**A uniform distribution is a good model for the outcome of rolling a fair die because each possible outcome has an equal chance of occurring.**

**When a fair die is rolled, the outcomes {1, 2, 3, 4, 5, 6} are all equally likely, with each number having a probability of 1/6. Since no outcome is more likely than another, the probabilities are evenly distributed across all possible values.**

**Therefore, the probability distribution is uniform, making the uniform distribution an appropriate and accurate model for a fair die.**

**8.How does log transformation help in stabilizing variance and making data more normally distributed?**

**Ans-**

**Log transformation is a data-preprocessing technique used to reduce skewness, stabilize variance, and make data more closely follow a normal distribution.**

1️⃣ Stabilizing Variance

In many real datasets (e.g., income, sales, population), variance increases with the mean.
Large values show much more spread than small values.

* Log transformation compresses large values more than small values, which:

* Reduces the effect of extreme values

* Makes variance more constant across the range of data (homoscedastic)

📌 Example:
Values: 10, 100, 1000
Log values: 1, 2, 3
→ Spread becomes more uniform

2️⃣ Making Data More Normally Distributed

Many datasets are right-skewed (long right tail).
Applying log transformation:

* Pulls in the long tail

* Makes the distribution more symmetric

* Brings it closer to a bell-shaped (normal) curve


3️⃣ Why It Is Important in Data Analysis

* Improves performance of statistical models that assume normality

* Helps linear regression meet assumptions

* Makes relationships more linear

* Reduces influence of outliers
